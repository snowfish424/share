# 12 — Elyra：檢查輸出 NC

**Pipeline Editor** 第 3 步（CPU）。確認 `CorrdiffOutput_*.nc` 存在。


```python
import os
from pathlib import Path

SHOME = os.environ.get("SHOME", "/mnt/corrdiff")
DATE = os.environ.get("DATE", "20260707")
out = Path(SHOME) / "dtg" / "EC_S2S_AIPP" / DATE / f"CorrdiffOutput_EC_RAW_{DATE}.nc"

print(f"=== check-output DATE={DATE} ===")
assert out.is_file(), f"Missing output: {out}"
print(f"[OK] {out} ({out.stat().st_size} bytes)")
try:
    import netCDF4 as nc
    with nc.Dataset(str(out)) as ds:
        print("[OK] Output NC dims:", dict(ds.dimensions))
except Exception as exc:
    print(f"[WARN] netCDF4 check skipped: {exc}")
print("check-output completed successfully.")
```
